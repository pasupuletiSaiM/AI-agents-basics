```python
import os
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-2.5-flash")
responce = model.invoke("who are you?")
responce.content
```

### Streaming and batch

```python
responce = model.stream("write a 50 words info on india")

for chunk in responce:
    print(chunk.content)
```

```python
responce2 = model.stream("write a 50 words info on india")
for chunk in responce2:
    print(chunk.content, end="~~", flush=True) 
```

### Batch

```python
MultiResponce = model.batch([
    "tell me about hyderabad culture",
    "why KCR lost in telangana elections?",
    "who is the first cm of AP?"
])

for chunk in MultiResponce:
    print(chunk.content, flush=True)
```

```python
MultiResponce = model.batch([
    "tell me about hyderabad culture",
    "why KCR lost in telangana elections?",
    "who is the first cm of AP?"
]
    config={
        'max_concurrency': 5   ## it will only process 5 responces at a time
    }
)

for chunk in MultiResponce:
    print(chunk.content, flush=True)
```

### Tools

```python
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """get the weather at a location"""
    return f"It's rainy at {location}"

model_with_tools = model.bind_tools([get_weather])
```

```python
responce = model_with_tools.invoke("what is the weather in hyderabad?")
responce
```
